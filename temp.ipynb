```python
import pyvista as pv
import numpy as np
from scipy.ndimage import binary_dilation

def extract_plane_points(biv, normal, center, offset_dist):
    """Return points from biv near the offset plane."""
    plane_origin = center - normal * offset_dist
    biv_points = biv.points
    distances = np.dot(biv_points - plane_origin, normal)
    plane_mask = np.abs(distances) <= tol
    distances_from_origin = np.linalg.norm(biv_points - plane_origin, axis=1)
    radius_mask = distances_from_origin <= (avg_tv_radius + radius_extra)
    mask = plane_mask & radius_mask
    return biv_points[mask]

for casenum in range(1,14): 
    print('==============')
    
    tvfile=f'/mnt/365a23c3-87b1-44cb-840c-37576193760e/Work/MS_HF_cohort/data/{casenum:02d}/1_seg/TV.surfmesh.vtk'
    bivfile=f'/mnt/365a23c3-87b1-44cb-840c-37576193760e/Work/MS_HF_cohort/data/{casenum:02d}/1_seg/RV_smooth.vtk'
    offset_dist = -20  # meters
    radius_extra = 15 # radial expansion in mm
    tol = 0.5  # meters

    outputfile=f'/mnt/365a23c3-87b1-44cb-840c-37576193760e/Work/MS_HF_cohort/data/{casenum:02d}/1_seg/TV_annulus{offset_dist}.vtk'
    print(outputfile)
    try: 
        # Load meshes
        tv = pv.read(tvfile)   # ring of points
        biv_init = pv.read(bivfile)      # larger mesh
        
        # Make sure ElemTag exists
        if "elemTag" not in biv_init.cell_data:
            raise ValueError("elemTag not found in cell_data.")

        # Get the ElemTag array
        elem_tag = biv_init.cell_data["elemTag"]

        # Create mask for cells where ElemTag == 3
        mask = elem_tag == 3

        # Extract those cells
        biv = biv_init.extract_cells(mask)
        
        # Get points from TV ring
        tv_points = tv.points

        # --- 1. Fit a plane to TV ring ---
        # centroid
        center = tv_points.mean(axis=0)

        # Calculate average TV radius
        distances_to_center = np.linalg.norm(tv_points - center, axis=1)
        #avgTVradius = np.mean(distances_to_center)
        # Average radius of TV ring
        avg_tv_radius = np.mean(np.linalg.norm(tv_points - center, axis=1))
        # Singular Value Decomposition to get normal
        _, _, vh = np.linalg.svd(tv_points - center)
        normal = vh[-1]  # plane normal vector

        # Ensure normal points in consistent direction (optional)
        normal = normal / np.linalg.norm(normal)

        # Try positive offset first
        plane_points = extract_plane_points(biv, normal, center, offset_dist)

        # If none, try negative offset
        if plane_points.size == 0:
            plane_points = extract_plane_points(biv, normal, center, -offset_dist)

        # Save result if found
        if plane_points.size > 0:
            plane_cloud = pv.PolyData(plane_points)
            plane_cloud.save(outputfile)
        else:
            print("No points found in either offset direction.")

        # Optional: visualize
        p = pv.Plotter()
        p.add_mesh(biv, color="lightblue", opacity=0.3)
        p.add_mesh(tv, color="green", point_size=10, render_points_as_spheres=True)
        p.add_mesh(plane_cloud, color="red", point_size=10, render_points_as_spheres=True)
        p.view_xz()

        p.show()
    except FileNotFoundError: 
        print(f'{outputfile} not created')
        continue
```

```python
def analyse_collfield_length_difference(df, collision_fields=['Micra'], collision_fields=['Combined_Collision', 'RV_Collision', 'TV_Collision', 'PM_Collision'], lengths=list(range(15, 50, 5))):
    """
    Perform statistical analysis comparing collision values between collision_fields and lengths.
    
    Parameters:
    -----------
    df : pandas.DataFrame
        DataFrame with collision data
    collision_fields : list
        List of collision fields to compare (default: =['Combined_Collision', 'RV_Collision', 'TV_Collision', 'PM_Collision'])
    lengths : list
        List of lengths to compare (default: [15, 20, 25, 30, 35, 40, 45])
    
    Returns:
    --------
    results : dict
        Dictionary containing statistical test results
    """
    
    results = {}
    
    # Filter data for specified collision_fields and lengths
    df_filtered = df[(df['collision_field'].isin(collision_fields)) & (df['length'].isin(lengths)) & (df['collision_field'].isin(collision_fields))]
    
    print(f"Analysis for collision_fields: {collision_fields}")
    print(f"Analysis for lengths: {lengths}")
    print(f"Total data points: {len(df_filtered)}")
    
    # Overall comparison (all regions combined)
    print("\n" + "="*80)
    print("OVERALL ANALYSIS (All AHA Regions Combined)")
    print("="*80)
    
    # Summary statistics by collision_field and length
    summary_stats = df_filtered.groupby(['collision_field', 'length'])['collision_value'].agg([
        'count', 'mean', 'std', 'median', 'min', 'max'
    ]).round(4)
    print("\nSummary Statistics by collision_field and Length:")
    print(summary_stats)
    
    # Summary statistics by collision_field only
    collision_field_stats = df_filtered.groupby('collision_field')['collision_value'].agg([
        'count', 'mean', 'std', 'median', 'min', 'max'
    ]).round(4)
    print("\nSummary Statistics by collision_field:")
    print(collision_field_stats)
    
    # Summary statistics by length only
    length_stats = df_filtered.groupby('length')['collision_value'].agg([
        'count', 'mean', 'std', 'median', 'min', 'max'
    ]).round(4)
    print("\nSummary Statistics by Length:")
    print(length_stats)

    # Two-way ANOVA (collision_field × length)
    print(f"\nTwo-way ANOVA (collision_field × Length):")
    try:
        import statsmodels.api as sm
        from statsmodels.formula.api import ols
        
        # Two-way ANOVA
        model = ols('collision_value ~ C(collision_field) + C(length) + C(collision_field):C(length)', 
                   data=df_filtered).fit()
        anova_table = sm.stats.anova_lm(model, typ=2)
        print(anova_table)
        
        results['overall'] = {
            'two_way_anova': anova_table.to_dict(),
            'collision_field_comparisons': {},
            'length_comparisons': {},
            'collision_field_length_combinations': {},
            'mixed_effects': {}
        }
    except Exception as e:
        print(f"  Error fitting two-way ANOVA: {e}")
        results['overall'] = {'error': str(e)}

    # collision_field comparisons (across all lengths)
    print(f"\ncollision_field Comparisons (across all lengths):")
    collision_field_groups = [df_filtered[df_filtered['collision_field'] == collision_field]['collision_value'].values 
                    for collision_field in collision_fields]
    
    # ANOVA for collision_fields
    f_stat_collision_field, p_anova_collision_field = stats.f_oneway(*collision_field_groups)
    print(f"ANOVA F-statistic: {f_stat_collision_field:.4f}")
    print(f"ANOVA p-value: {p_anova_collision_field:.6f}")
    print(f"Significant difference: {'Yes' if p_anova_collision_field < 0.05 else 'No'}")
    
    # Pairwise t-tests between collision_fields
    print(f"\nPairwise t-tests between collision_fields:")
    for collision_field1, collision_field2 in combinations(collision_fields, 2):
        group1 = df_filtered[df_filtered['collision_field'] == collision_field1]['collision_value'].values
        group2 = df_filtered[df_filtered['collision_field'] == collision_field2]['collision_value'].values
        
        if len(group1) > 0 and len(group2) > 0:
            t_stat, p_val = stats.ttest_ind(group1, group2)
            results['overall']['collision_field_comparisons'][f'{collision_field1}_vs_{collision_field2}'] = {
                't_stat': t_stat, 'p_value': p_val
            }
            
            print(f"  {collision_field1} vs {collision_field2}: t={t_stat:.4f}, p={p_val:.6f} {'*' if p_val < 0.05 else ''}")

    # Length comparisons (across all collision_fields)
    print(f"\nLength Comparisons (across all collision_fields):")
    length_groups = [df_filtered[df_filtered['length'] == length]['collision_value'].values 
                    for length in lengths]
    
    # ANOVA for lengths
    f_stat_length, p_anova_length = stats.f_oneway(*length_groups)
    print(f"ANOVA F-statistic: {f_stat_length:.4f}")
    print(f"ANOVA p-value: {p_anova_length:.6f}")
    print(f"Significant difference: {'Yes' if p_anova_length < 0.05 else 'No'}")
    
    # Pairwise t-tests between lengths
    print(f"\nPairwise t-tests between lengths:")
    for length1, length2 in combinations(lengths, 2):
        group1 = df_filtered[df_filtered['length'] == length1]['collision_value'].values
        group2 = df_filtered[df_filtered['length'] == length2]['collision_value'].values
        
        if len(group1) > 0 and len(group2) > 0:
            t_stat, p_val = stats.ttest_ind(group1, group2)
            results['overall']['length_comparisons'][f'{length1}_vs_{length2}'] = {
                't_stat': t_stat, 'p_value': p_val
            }
            
            print(f"  {length1} vs {length2}: t={t_stat:.4f}, p={p_val:.6f} {'*' if p_val < 0.05 else ''}")

    # Mixed Effects Model
    print(f"\nMixed Effects Model (collision_field + length = fixed, region = random):")
    try:
        model = smf.mixedlm("collision_value ~ C(collision_field) + C(length)", 
                           df_filtered, groups=df_filtered["AHA_region"])
        mixed_result = model.fit()
        print(mixed_result.summary())
        
        results['overall']['mixed_effects']['summary'] = mixed_result.summary().as_text()
        results['overall']['mixed_effects']['pvalues'] = mixed_result.pvalues.to_dict()
    except Exception as e:
        print(f"  Error fitting mixed effects model: {e}")
        results['overall']['mixed_effects']['error'] = str(e)

    # Analysis by collision_field (comparing lengths within each collision_field)
    print("\n" + "="*80)
    print("ANALYSIS BY collision_field (comparing lengths within each collision_field)")
    print("="*80)
    
    results['by_collision_field'] = {}
    for collision_field in collision_fields:
        print(f"\ncollision_field: {collision_field}")
        collision_field_data = df_filtered[df_filtered['collision_field'] == collision_field]
        
        if len(collision_field_data) == 0:
            print("  No data available")
            continue
        
        # Summary stats for this collision_field
        collision_field_length_stats = collision_field_data.groupby('length')['collision_value'].agg([
            'count', 'mean', 'std', 'median'
        ]).round(4)
        print(collision_field_length_stats)
        
        # ANOVA for lengths within this collision_field
        collision_field_length_groups = [collision_field_data[collision_field_data['length'] == length]['collision_value'].values 
                               for length in lengths]
        non_empty_groups = [g for g in collision_field_length_groups if len(g) > 0]
        
        if len(non_empty_groups) < 2:
            print("  Insufficient data for comparison")
            continue
        
        f_stat_collision_field_length, p_anova_collision_field_length = stats.f_oneway(*non_empty_groups)
        results['by_collision_field'][collision_field] = {
            'anova': {'f_stat': f_stat_collision_field_length, 'p_value': p_anova_collision_field_length},
            'pairwise': {}
        }
        
        print(f"  ANOVA: F={f_stat_collision_field_length:.4f}, p={p_anova_collision_field_length:.6f} {'*' if p_anova_collision_field_length < 0.05 else ''}")
        
        # Pairwise comparisons for lengths within this collision_field
        available_lengths = collision_field_data['length'].unique()
        if len(available_lengths) >= 2:
            print(f"  Pairwise length comparisons:")
            for length1, length2 in combinations(available_lengths, 2):
                group1 = collision_field_data[collision_field_data['length'] == length1]['collision_value'].values
                group2 = collision_field_data[collision_field_data['length'] == length2]['collision_value'].values
                
                if len(group1) > 0 and len(group2) > 0:
                    t_stat, p_val = stats.ttest_ind(group1, group2)
                    results['by_collision_field'][collision_field]['pairwise'][f'{length1}_vs_{length2}'] = {
                        't_stat': t_stat, 'p_value': p_val
                    }
                    
                    print(f"    {length1} vs {length2}: t={t_stat:.4f}, p={p_val:.6f} {'*' if p_val < 0.05 else ''}")

    # Analysis by Length (comparing collision_fields within each length)
    print("\n" + "="*80)
    print("ANALYSIS BY LENGTH (comparing collision_fields within each length)")
    print("="*80)
    
    results['by_length'] = {}
    for length in lengths:
        print(f"\nLength: {length}")
        length_data = df_filtered[df_filtered['length'] == length]
        
        if len(length_data) == 0:
            print("  No data available")
            continue
        
        # Summary stats for this length
        length_collision_field_stats = length_data.groupby('collision_field')['collision_value'].agg([
            'count', 'mean', 'std', 'median'
        ]).round(4)
        print(length_collision_field_stats)
        
        # ANOVA for collision_fields within this length
        length_collision_field_groups = [length_data[length_data['collision_field'] == collision_field]['collision_value'].values 
                               for collision_field in collision_fields]
        non_empty_groups = [g for g in length_collision_field_groups if len(g) > 0]
        
        if len(non_empty_groups) < 2:
            print("  Insufficient data for comparison")
            continue
        
        f_stat_length_collision_field, p_anova_length_collision_field = stats.f_oneway(*non_empty_groups)
        results['by_length'][length] = {
            'anova': {'f_stat': f_stat_length_collision_field, 'p_value': p_anova_length_collision_field},
            'pairwise': {}
        }
        
        print(f"  ANOVA: F={f_stat_length_collision_field:.4f}, p={p_anova_length_collision_field:.6f} {'*' if p_anova_length_collision_field < 0.05 else ''}")
        
        # Pairwise comparisons for collision_fields within this length
        available_collision_fields = length_data['collision_field'].unique()
        if len(available_collision_fields) >= 2:
            print(f"  Pairwise collision_field comparisons:")
            for collision_field1, collision_field2 in combinations(available_collision_fields, 2):
                group1 = length_data[length_data['collision_field'] == collision_field1]['collision_value'].values
                group2 = length_data[length_data['collision_field'] == collision_field2]['collision_value'].values
                
                if len(group1) > 0 and len(group2) > 0:
                    t_stat, p_val = stats.ttest_ind(group1, group2)
                    results['by_length'][length]['pairwise'][f'{collision_field1}_vs_{collision_field2}'] = {
                        't_stat': t_stat, 'p_value': p_val
                    }
                    
                    print(f"    {collision_field1} vs {collision_field2}: t={t_stat:.4f}, p={p_val:.6f} {'*' if p_val < 0.05 else ''}")

    # Analysis by AHA Region
    print("\n" + "="*80)
    print("ANALYSIS BY AHA REGION")
    print("="*80)
    
    results['by_region'] = {}
    for region in sorted(df_filtered['AHA_region'].unique()):
        print(f"\nAHA Region {region}:")
        region_data = df_filtered[df_filtered['AHA_region'] == region]
        
        if len(region_data) == 0:
            print("  No data available")
            continue
        
        # Summary stats by collision_field and length within region
        region_stats = region_data.groupby(['collision_field', 'length'])['collision_value'].agg([
            'count', 'mean', 'std'
        ]).round(4)
        print(region_stats)
        
        # Two-way ANOVA within region (if sufficient data)
        try:
            if len(region_data) > 10:  # Minimum threshold for meaningful analysis
                model_region = ols('collision_value ~ C(collision_field) + C(length) + C(collision_field):C(length)', 
                                 data=region_data).fit()
                anova_table_region = sm.stats.anova_lm(model_region, typ=2)
                print(f"  Two-way ANOVA for region {region}:")
                print(anova_table_region)
                
                results['by_region'][region] = {
                    'two_way_anova': anova_table_region.to_dict(),
                    'sample_size': len(region_data)
                }
            else:
                print(f"  Insufficient data for ANOVA (n={len(region_data)})")
                results['by_region'][region] = {'insufficient_data': len(region_data)}
        except Exception as e:
            print(f"  Error in regional analysis: {e}")
            results['by_region'][region] = {'error': str(e)}
    
    return results
```
