```python
# Make cylinder mesh 
import pyvista as pv
import numpy as np

path="/data_fast/Work/MS_HF_cohort/data/01/"

# Parameters
import numpy as np
devices={
    "Micra": {"D": 6.7, "L": 25.9}, # AV/AV2
    "AveirAR": {"D": 6.5, "L": 32.2}, # atrial Aveir DR device 
    "AveirVR": {"D": 6.5, "L": 38} # Atrial Aveir DR device + 10mm
}

#for dev in ["AveirVR", "AveirAR" , "Micra"]:
dev="AveirVR"
D = devices[dev]["D"]
L = devices[dev]["L"]
R = D / 2

target_tags = [4]  # The elemTags you're interested in

# Load meshes
for time in range(0,11):
    mesh2 = pv.read(path+"2_motion/tracked/transformed-{}.vtk".format(time)) # Is the transformed data
    mesh1 = pv.read(path+"2_motion/tracked/RV_septum-{}.vtk".format(time)) # Mesh with normals
    # Step 1: Filter mesh2 based on elemTags
    if "elemTag" not in mesh2.cell_data:
        raise ValueError("mesh2 does not contain 'elemTag' in cell_data")

    # mask = mesh2.cell_data['elemTag'] == 40
    # mesh1=mesh2.extract_cells(mask) # Corresponds to the mesh with the sparse points to anchor the cylinder + normals on this mesh
    mesh1=mesh1.extract_surface() 
    # tags = mesh2.cell_data["elemTag"]
    # mask = np.isin(tags, target_tags)
    # filtered_mesh2 = mesh2.extract_cells(mask)

    # Step 2: Get normals from mesh1
    # normals = mesh1.point_data.get("Normals")
    # if normals is None:
    #     raise ValueError("mesh1 does not contain 'Normals'")
    mesh1.compute_normals(inplace=True)

    # N = 1000
    # new_points   = mesh1.points[::N]
    # new_normals  = mesh1.point_data["Normals"][::N]
    N = 3000 # Mid RV septum
    #N = 3763 # PM and RV wall
    #N = 5623 # TV 6330 #1822 #
    # LAF
    N=62063
    new_points   = mesh1.points[[N]]
    new_normals  = mesh1.point_data["Normals"][[N]]
    # Flip normals to point outward from the RV septum
    new_normals = -new_normals
    print(f'Number of cylinders to plot: {len(new_points)}')
    index=0
    for point, normal in zip(new_points, new_normals):
        
        normal = normal / np.linalg.norm(normal)
        print(index)
        print(f'Point: {point}')
        print(f'Normal: {normal}')
        # Center the cylinder
        center = point + (normal * (L / 2.0))

        # Build the cylinder
        cylinder = pv.Cylinder(center=center, direction=normal, radius=R, height=L)
        cylinder_output_path=f'/data_fast/Work/MS_HF_cohort/data/01/2_motion/tracked/{dev}_cylinder{N}_time{time}.vtp'
        cylinder.save(cylinder_output_path)
        # plotter = pv.Plotter()
        # #plotter.add_mesh(mesh2, color='lightgrey', opacity=0.5, show_edges=True)
        # plotter.add_mesh(cylinder, color='red', opacity=0.8)
        # plotter.add_mesh(mesh1, color='blue', opacity=0.5)
        # # Rotate for better view
        # plotter.camera_position = 'yz'  
        # plotter.show()
        index+=1
```
